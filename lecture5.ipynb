For loop

for loops are seqential traversal. For traversing list,string,tuples etc.

```python
nums = [1,4,9,16,25,36,49,64,81,100]
for n in nums :
    print(n)
```

```python
nums = (1,4,9,16,25,36,49,64,81,100)
x = 49 
for n in nums :
    if (n==x):
        print("found it ",x)
        break
    else:
        print("x is  not found")
```
