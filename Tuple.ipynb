```python
#Tuples
#Tuple is similar to list

#empty tuples
t = ()

#tuple having integers
t = (1,2,3,4,5)
print(t)

# mixed tuple  
t = (1, 'raju', 28 ,'geet')
print(t)
#nested tuple
t = (1,(2,3,4),[1,'raju',2,'geet',])
print(t)
```

```python
#concating tuples
t = (1,2,3) + (4,5,6)
print(t)
```

```python
#only parenthesis not enough
t = ('satish',)
print(t)
```

```python
#parethesis are optional
t= "satish",
print(type(t))
print(t)
```

```python
"""accesing elemengts in tuple"""
t = ('ram','shyam','kapil','raj')
print(t[1])
"""calling last element"""
print(t[-1])
```

```python
t = ('abc',('satish','karn','nitin'))
print(t[-1][0])
```

```python
#slicing
t = (1,2,3,4,5,6)
print(t[1:4])

#from starting to 2nd last element
print(t[:-2])

#from starting to end
print(t[:])
```

```python
#creating tuple
t = (1,2,3,4, [5,6,7])
t[2] = 'x' #will get Type error 
```

```python
t = ('abc',('satish','karn','nitin'))
t[4][2] = 'jitendra'
print(t)
```

```python
#repeat the elements in a tuple for a given number of times
t = (('satish',) * 10)
print(t)
```
