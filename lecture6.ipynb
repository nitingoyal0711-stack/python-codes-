Recursion


function call it self again and again 

```python
#recursion

def show(n):

    if (n==0):#base case
        return
    print(n)
    show(n-1)
show(6)    
```

```python
def show(n):
    if (n == 0):

        return
    print(n)
    show(n-1)
    print("END")     
show(3)
```

```python
#to print all the all the natural number by recursion
def print_nat(n):
    if (n == 0):
        return
    print(n)
    print_nat(n-1)
print_nat(5)
```
