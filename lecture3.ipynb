Q2: wap to check if a list contains a palindrome(when we read both directions {bakward & forward} , we get the same results ) of elements .(hint : use copy() method)

```python
list = [1,2,3]
copy_list1 = list.copy()

copy_list1.reverse() 

if (copy_list1 == list):
    print("palindrom of elements")
else:
    print("non-palindromic list ")   
    print(list) 
```
